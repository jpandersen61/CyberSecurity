### Exercise 3.4.1: Create a Dictionary

```python
movie = {
    "title": "Inception",
    "director": "Christopher Nolan",
    "year": 2010,
    "rating": 8.8
}

print(movie)
```

### Exercise 3.4.2: Accessing and Modifying Dictionary Values

*Code block:*

```python
print(movie["title"])

movie["rating"] = 9.0

print(movie)
```

### Exercise 3.4.3: Iterating Over a Dictionary

*Code block:*
