###**Exercise:** Converting decimal (integer) to binary

Make a Python script that converts from decimal to binary - e.g. 20 -> 10100. The script should prompt the user for the number to be converted with the [input function](https://docs.python.org/3/builtins/functions.html) and print out the result.

Hint: You can do the conversion with the [format function](http://docs.python.org/3/builtins/functions.html) like this `format(decimal, 'b')`

### **Solution**: Converting decimal (integer) to binary

Use the 'b' [integer presentation type specifier:](https://docs.python.org/3/library/string.html#formatspec) applying to the `format()` function to convert to binary. As an extra we will convert to hexadecimal as well with the 'x' specifier.

```python
#Tell user that we are going to convert decimal to binary -> print message on screen
print("Converting decimal to binary")

#Prompt user to enter a number
number  = int(input("Enter number (decimal): "))

#Leave an empty line before printing the result
print()

#Convert number to binary as string type/class (str)
binary = format(number, 'b')

#Print the result
print(f"Binary of {number} is '{binary}'")
print(f"Converted a variable og type {type(binary)}")

#Leave an empty line before printing the result
print()

#Convert number to hexadecimal as string type/class (str)
hex = format(number, 'x')

#Print the result
print(f"Hexadecimal of {number} is '{hex}'")
print(f"Converted a variable of type: {type(hex)}")

```

###**Exercise**: Converting from binary to decimal

Make a Python script that converts from binary to decimal - e.g. 10100 -> 20. The script should prompt the user for the number to be converted with the [input function](https://docs.python.org/3/builtins/functions.html) and print out the result.

Hint: You can do the convertion with the [int function](http://docs.python.org/3/builtins/functions.html) like this `int(binary, 2)`

###**Solution**: Converting from binary to decimal

```python
#Tell user that we are going to convert binary to decimal -> print message on screen
print("Converting binary to decimal")

#Prompt user to enter a number
binary  = input("Enter number (binary): ")

#Leave an empty line before printing the result
print()

#Convert binary to decimal as string type/class (int)
decimal = int(binary, 2)

#Print the result
print(f"Decimal value of {binary} is {decimal}")
print(f"Converted a variable of type: {type(decimal)}")
```
